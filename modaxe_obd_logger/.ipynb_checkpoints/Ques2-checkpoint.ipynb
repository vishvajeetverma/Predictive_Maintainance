```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
img = cv2.imread('../buildings.jpg', cv2.IMREAD_COLOR_RGB)
plt.imshow(img)
plt.axis('off')
plt.show()
print("Image shape:", img.shape, "min/max:", img.min(), img.max())
```

### Question 2a

```python
def get_gaussian_kernel(k : int, sigma : float):
    center = k//2
    kernel = np.zeros((k,k))
    for i in range(k) :
        for j in range(k) :
            kernel[i][j] = np.exp(-((i - center)**2 + (j-center)**2) / (2*(sigma**2)))
    kernel /= np.sum(kernel)
    return kernel

def get_padded_gaussian_kernel(kernel, H, W) :
    kh, kw = kernel.shape

    padded_kernel = np.zeros((H, W))
    center_i = (H - kh) // 2
    center_j = (W - kw) // 2
    padded_kernel[center_i:center_i + kh, center_j:center_j + kw] = kernel

    return np.fft.ifftshift(padded_kernel)
```

```python
k, sigma = 13, 2.5
kernel = get_gaussian_kernel(k, sigma)

H, W, C = img.shape
padded_kernel = get_padded_gaussian_kernel(kernel, img.shape[0], img.shape[1])

blurred = []
kernel_dft = np.fft.fft2(padded_kernel)

for c in range(C) :
    img_dft = np.fft.fft2(img[:,:,c])
    blur = img_dft * kernel_dft
    blur = np.real(np.fft.ifft2(blur))
    blurred.append(blur)

blurred_img = np.stack(blurred, axis = 2).astype(np.uint8)

fig, ax = plt.subplots(1,3, figsize=(12, 5))

ax[0].imshow(img)
ax[0].axis('off')
ax[0].set_title("Original Image")

ax[1].imshow(kernel)
ax[1].set_title("Centered Gaussian Kernel (13 x 13, sigma = 2.5)")
# ax[1].colorbar(label='Value')

ax[2].imshow(blurred_img)
ax[2].axis('off')
ax[2].set_title("Blurred Image")

plt.tight_layout()
plt.show()
```

### Question 2b

```python
k1, k2, sigma = 13, 1036, 2.5
fig, ax = plt.subplots(2,2, figsize = (12,6))
n = 1
kernel = get_gaussian_kernel(k1,sigma)
kernel_c = np.fft.ifftshift(kernel) 
kernel_dft = np.fft.fftshift(np.fft.fft2(kernel_c))

kernel_dft_real = np.abs(kernel_dft)
ax[0,0].imshow(kernel_dft_real)
ax[0,0].set_title("2D Centered Gaussian Magnitude Spectrum (13x13)")


epsilon = 1e-3
inverse_kernel_dft = 1/(kernel_dft_real + epsilon)
ax[0,1].imshow(inverse_kernel_dft)
ax[0,1].set_title("2D Inverse Centered Magnitude Spectrum (13x13)")

kernel_2 = get_gaussian_kernel(k2,sigma)
kernel_2_c = np.fft.ifftshift(kernel_2) 
kernel_2_dft = np.fft.fftshift(np.fft.fft2(kernel_2_c))

kernel_2_dft_real = np.abs(kernel_2_dft)
ax[1,0].imshow(kernel_2_dft_real)

ax[1,0].set_title("2D Centered Gaussian Magnitude Spectrum (1036x1036)")


epsilon = 1e-3
kernel_2_kernel_dft = 1/(kernel_2_dft_real + epsilon)
ax[1,1].imshow(kernel_2_kernel_dft)
ax[1,1].set_title("2D Inverse Centered Magnitude Spectrum (1036x1036)")

plt.tight_layout()
plt.show()
```

### Question 2c

```python
def get_H_cont(M, N, k):
    u = np.arange(M) - (M - 1) / 2
    v = np.arange(N) - (N - 1) / 2
    U, V = np.meshgrid(u, v, indexing='ij')
    return np.exp(-k * (U**2 + V**2))
```

```python
M, N = 1036, 1036
H = get_gaussian_kernel(M,sigma)
H_c = np.fft.ifftshift(H) 
H_DFT = np.abs(np.fft.fftshift(np.fft.fft2(H_c)))

from tqdm import tqdm
k_values = np.linspace(1e-6, 1e-3, 2000)
min_error = float('inf')
opt_k = None
error_log = []
for k in tqdm(k_values) :
    H_cont = get_H_cont(M, N, k)
    error = np.sum((H_cont - H_DFT)**2)
    # print(approx)
    error_log.append(error)
    if error < min_error :
        min_error = error
        opt_k = k

print("Optimal k : ",opt_k)

```

```python
plt.figure(figsize=(8,5))
plt.plot(k_values, error_log, color='blue', linewidth=2, label='Error')
plt.xlabel('k (x2000)', fontsize=12)
plt.yscale('log')
plt.ylabel('Error(log scale)', fontsize=12)
plt.axvline(x=opt_k, color='red', linestyle='--', linewidth=1)
plt.axhline(y=min(error_log), color='green', linestyle='--', linewidth=1)
plt.title('Error vs k for Continuous Gaussian Filter Approximation', fontsize=12)
plt.grid(linestyle='--', alpha=0.7)
plt.show()
```

```python
H_opt = get_H_cont(M,N, opt_k)

print("Optimal K : ", opt_k)
fig, ax = plt.subplots(1,2, figsize = (12,9))
ax[0].imshow(H_opt)
ax[0].set_title("Magintude spectrum of Gaussian fit |H_cont|")

ax[1].imshow(1/(H_opt + epsilon))
ax[1].set_title("Inverse Magnitude spectrum of Gaussian fit")

plt.tight_layout()
plt.show()
```

### Question 2D

```python
def get_new_img(img, kernel_dft) :
    _, _, C = img.shape
    # print(C)
    new_img = []
    for c in range(C) :
        img_dft = np.fft.fft2(img[:,:,c])
        blur = img_dft * kernel_dft
        blur = np.real(np.fft.ifft2(blur))
        new_img.append(blur)

    new_img = np.stack(new_img, axis = 2).astype(np.uint8)
    return new_img

def get_mse(img_1, img_2) :
    H,W,C = img_1.shape
    return np.mean((img_1 - img_2)**2)
```

```python
pad = 6

# Define border type and color (e.g., BORDER_CONSTANT with blue color)
padded_img = cv2.copyMakeBorder(img, pad, pad, pad, pad, 
                                cv2.BORDER_CONSTANT, value=[0, 0, 0])
opt_k = 0.000115
M,N, C = padded_img.shape
H = get_gaussian_kernel(M,sigma)
H_c = np.fft.fftshift(H) 
H_DFT = np.fft.fft2(H_c)

H_cont = get_H_cont(M,N, opt_k)
H_cont = np.fft.fftshift(H_cont)
recon_img_1 = get_new_img(padded_img, H_DFT)
recon_img_2 = get_new_img(padded_img, H_cont)

fig, ax = plt.subplots(1,3, figsize = (12,9))
ax[0].imshow(padded_img)
ax[0].set_title("Original Image")

ax[1].imshow(recon_img_1)
ax[1].set_title("Reconstructed Image (Direct Kernel Inverse)")

ax[2].imshow(recon_img_2)
ax[2].set_title("Reconstructed Image (Gaussian Fit Inverse)")

plt.tight_layout()
plt.show()

print(f"MSE b/w original Image and Reconstructed Image using Direct Kernel Inverse : {get_mse(padded_img, recon_img_1)} ")
print(f"MSE b/w original Image and Reconstructed Image using Gaussian Fit Inverse : {get_mse(padded_img, recon_img_2)} ")
```
